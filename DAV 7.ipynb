```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.preprocessing import LabelEncoder

# =================================================================
# SECTION 1: DATA PREPARATION & CLEANING
# =================================================================
# Setting seed for reproducible dataset simulation
np.random.seed(42)

# Generate Mock UCI Dataset
uci_df = pd.DataFrame({
    "Age": np.random.randint(20, 70, size=100),
    "Class": np.random.choice(["Positive", "Negative"], size=100)
})

# Generate Mock Pima Dataset
pima_df = pd.DataFrame({
    "Glucose": np.random.randint(70, 200, size=100),
    "BMI": np.random.uniform(18.0, 45.0, size=100),
    "Outcome": np.random.choice([0, 1], size=100)
})

# Clean clinical zeroes in Pima Dataset
pima_df["Glucose"] = pima_df["Glucose"].replace(0, np.nan)
pima_df["BMI"] = pima_df["BMI"].replace(0, np.nan)
pima_df = pima_df.dropna()

# Encode the text labels in the UCI Dataset to 0 and 1 for logistic regression
le = LabelEncoder()
uci_df["Class_Encoded"] = le.fit_transform(uci_df["Class"])


# =================================================================
# SECTION 2: LINEAR REGRESSION (Continuous vs. Continuous)
# =================================================================
print("================ BIVARIATE LINEAR REGRESSION ================")

# Problem: How does Glucose predict BMI in the Pima Indians Dataset?
X_lin = pima_df[["Glucose"]]
y_lin = pima_df["BMI"]

lin_model = LinearRegression()
lin_model.fit(X_lin, y_lin)

print("--- Pima Indians Dataset: Glucose vs BMI ---")
print(f"Intercept (Alpha) : {lin_model.intercept_:.4f}")
print(f"Slope (Beta Coeff): {lin_model.coef_[0]:.4f}")
print(f"R-squared Score   : {lin_model.score(X_lin, y_lin):.4f}")
print("Interpretation: For every 1-unit increase in Glucose, BMI changes by the slope value.")


# =================================================================
# SECTION 3: LOGISTIC REGRESSION (Continuous vs. Binary Target)
# =================================================================
print("\n================ BIVARIATE LOGISTIC REGRESSION ================")

# 1. UCI Dataset Analysis: Does Age predict the Class outcome?
X_log_uci = uci_df[["Age"]]
y_log_uci = uci_df["Class_Encoded"]

log_model_uci = LogisticRegression()
log_model_uci.fit(X_log_uci, y_log_uci)

print("--- UCI Dataset: Age vs Class ---")
print(f"Intercept         : {log_model_uci.intercept_[0]:.4f}")
print(f"Beta Coefficient  : {log_model_uci.coef_[0][0]:.4f}")
print(f"Odds Ratio        : {np.exp(log_model_uci.coef_[0][0]):.4f}")

# 2. Pima Dataset Analysis: Does Glucose predict the Diabetes Outcome?
X_log_pima = pima_df[["Glucose"]]
y_log_pima = pima_df["Outcome"]

log_model_pima = LogisticRegression()
log_model_pima.fit(X_log_pima, y_log_pima)

print("\n--- Pima Indians Dataset: Glucose vs Outcome ---")
print(f"Intercept         : {log_model_pima.intercept_[0]:.4f}")
print(f"Beta Coefficient  : {log_model_pima.coef_[0][0]:.4f}")
print(f"Odds Ratio        : {np.exp(log_model_pima.coef_[0][0]):.4f}")
print("Interpretation: An Odds Ratio > 1 implies the risk increases as the predictor variable grows.")
```
